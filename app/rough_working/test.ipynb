```python
#!/usr/bin/env python3
"""
Replace 'accepted_converted_to_jpg' with 'accepted' in the 'file_path' field
of each record in a JSONL file. The original file is left unchanged.
"""

import json
from pathlib import Path
import argparse

def replace_file_path(input_file: Path, output_file: Path):
    count_total = 0
    count_modified = 0

    with input_file.open("r", encoding="utf-8") as fin, \
         output_file.open("w", encoding="utf-8") as fout:
        for line_no, line in enumerate(fin, 1):
            line = line.strip()
            if not line:
                continue
            try:
                obj = json.loads(line)
                count_total += 1

                # Perform replacement only if key exists and contains the target string
                if "file_path" in obj and "accepted_converted_to_jpg" in obj["file_path"]:
                    obj["file_path"] = obj["file_path"].replace("accepted_converted_to_jpg", "accepted")
                    count_modified += 1

                fout.write(json.dumps(obj, ensure_ascii=False) + "\n")

            except json.JSONDecodeError as e:
                print(f"⚠️ Skipping bad JSON line {line_no}: {e}")

    print(f"\n✅ Done. Processed {count_total} lines; modified {count_modified} file_path entries.")
    print(f"➡️ Output saved to: {output_file.resolve()}")
```

```python
input_path = Path("output_filter1/accepted/metadata_jpg.jsonl")         # or your file name
output_path = Path("output_filter1/accepted/metadata_fixed.jsonl")  # choose your new name
replace_file_path(input_path, output_path)
```
